```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Load the generated results
try:
    df = pd.read_csv("../outputs/raw/results_raw.csv")
    display(df.head())
except FileNotFoundError:
    print("Results not found. Please run src/run.py first.")
```

## 📊 High-Level Metrics
Comparing overall latency and cost.

```python
if 'df' in locals():
    summary = df.groupby('Model').agg({
        'Latency_sec': 'mean',
        'Estimated_Cost_USD': 'sum',
        'Total_Tokens': 'sum'
    }).reset_index()
    display(summary)
    
    # Plot Latency
    plt.figure(figsize=(8,4))
    plt.bar(summary['Model'], summary['Latency_sec'], color=['#2ca02c', '#d62728'])
    plt.title('Average Latency by Model')
    plt.ylabel('Seconds')
    plt.show()
```
